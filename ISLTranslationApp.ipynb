```python
import tkinter as tk
import time
import tensorflow as tf
import cv2
import threading
import numpy as np
```

```python
model = tf.keras.models.load_model(r'Models\islModel.keras')
vidInput = cv2.VideoCapture(0)
root = tk.Tk()
# capturedFrames = []
isCapturing = False
currentLetter = False
previousLetter = False
translatedLetters = ''
```

```python
def captureVideo():
    global root
    global capturedFrames
    global currentLetter
    global previousLetter
    global translatedText
    global translatedLetters
    
    vidCanvas = tk.Canvas(root, width = 640, height = 480)
    vidCanvas.pack()
    
    while isCapturing:
        ret, frame = vidInput.read()
        if ret:
            frame = cv2.resize(frame, (224, 224))
            frame = np.expand_dims(frame, axis=0)

#             capturedFrames.append(frame)
            cv2.imshow('Video', frame)
            
            if cv2.waitKey(1) & 0xFF == ord('q'):
                break
        
        currentLetter = model.predict(frame)
        
        if currentLetter != None:
            if currentLetter != previousLetter:
                translatedLetters += currentLetter
                translatedText['Text'] = translatedLetters
                previousLetter = currentLetter
        
    vidInput.release()
    cv2.destroyAllWindows()
```

```python
def createStopButton(item, win, can):
    item.destroy()
    
#     Stop button
    stopIcon = tk.PhotoImage(file = 'Images/stopButton.png')
    stopButton = tk.Button(win, image = stopIcon, bg = '#424242', bd = 0, command = lambda: stopCommand(stopButton, win, can))
    can.create_window(win.winfo_screenwidth()/1.7, win.winfo_screenheight()/2, window = stopButton)
    stopButton.pack()
    
#     Video capture
    global isCapturing
    isCapturing = True
    captureThread = threading.Thread(target=captureVideo)
    captureThread.start()
```

```python
def stopCommand(item, win, can):
    global isCapturing
    isCapturing = False
    item.destroy()
    
#     Record Video Button
    recIcon = tk.PhotoImage(file = 'Images/videoCaptureButton.png')
    recButton = tk.Button(win, image = recIcon, bg = '#424242', bd = 0)
    can.create_window(win.winfo_screenwidth()/1.7, win.winfo_screenheight()/2, window = recButton)    
```

```python
def translateButtonCommands():
    signLanguageToText()

def homeButtonCommands(root, win):
    win.destroy()
    win.quit()
    home(root)

def recordVideoButtonCommands(item, win, can):
    createStopButton(item, win, can)
```

```python
# Home window   
def home(root):
    canvas = tk.Canvas(root, width = root.winfo_screenwidth(), height = root.winfo_screenheight())
    canvas.pack()
    canvas['bg'] = '#424242'

#     Title
    title = tk.Label(root, text = 'Translate Sign Language', font = ('Helvetica', 48), bg = '#424242', fg = '#ffffff')
    canvas.create_window(root.winfo_screenwidth()/2, root.winfo_screenheight()/9, window = title)

#     Translate Button
    translate = tk.PhotoImage(file = 'Images/translateButtonDefault.png')
    translateButton = tk.Button(root, image = translate, bg = '#424242', bd = 0, command = lambda: translateButtonCommands())
    canvas.create_window(root.winfo_screenwidth()/2, root.winfo_screenheight()/2.75, window = translateButton)
    
#     Mainloop
    root.mainloop()
```

```python
# Sign Language to Text Screen
def signLanguageToText():
    SLT = tk.Toplevel()
    SLTcanvas = tk.Canvas(SLT, width = SLT.winfo_screenwidth(), height = SLT.winfo_screenheight())
    SLTcanvas.pack()
    SLTcanvas['bg'] = '#424242'
    
#     Home Button
    homeIcon = tk.PhotoImage(file = 'Images/homeButton.png')
    homeButton = tk.Button(SLT, image = homeIcon, bg = '#424242', bd = 0, command =  lambda: homeButtonCommands(root, SLT))
    SLTcanvas.create_window(SLT.winfo_screenwidth()/2.4, SLT.winfo_screenheight()/2, window = homeButton)
    
#     Record Button
    recIcon = tk.PhotoImage(file = 'Images/videoCaptureButton.png')
    recButton = tk.Button(SLT, image = recIcon, bg = '#424242', bd = 0, command = lambda: recordVideoButtonCommands(recButton, SLT, SLTcanvas))
    SLTcanvas.create_window(SLT.winfo_screenwidth()/1.7, SLT.winfo_screenheight()/2, window = recButton)
    
#     translated text
    translatedText = tk.Label(SLT, text = '', font = 'Helvetica', bg = '#424242', fg = '#ffffff')
    SLTcanvas.create_window(SLT.winfo_screenwidth()/10, SLT.winfo_screenheight()/2.3, window = translatedText)
    
#     Mainloop
    SLT.mainloop()
```

```python
home(root)
```
